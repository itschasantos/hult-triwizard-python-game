<b> Welcome to the <i>TriWizard Quest<i/></b><b>! A thrilling adventure game set in the magical world of Harry Potter. If you are not a Harry Potter fan, worry not, this game will be as much fun :) </b>

This game will take you to a journey through the infamous Triwizard Tournament, an event that challenges wizards and witches with tasks of bravery, intelligence, and magic. This historic contest brings together three prestigious magical schools: Hogwarts School of Witchcraft and Wizardry, Beauxbatons Academy of Magic, and Durmstrang Institute. These institutions compete for glory through their chosen champions facing a quest that test their limits.

The champions, selected as the finest from their schools, enter a fierce competition to earn the title of the best. To ensure fairness, students submit their names to the mystical Goblet of Fire, which impartially selects the most deserving student from each school and crown them as their champion.

This year, however, the Goblet of Fire has made an unexpected choice. In addition to the usual champions, it has selected an extra contender from Hogwarts – You!!! While this may seem scary, rest assured, you're not alone. A supportive network from your chosen house will assist you in conquering the tournament's three quests. 

The path ahead is loaded with challenges, and each decision you make will significantly influence your journey. Will you rise to the occasion, proving your worth in this historic competition, or will the trials prove too challenging? Proving those who doubt the Goblet of Fire choosing you, right? The future of Hogwart's name in the tournament rests in your hands.

<b> Choose your Hogwarts house wisely, be strategic, and bravely step into the arena of <i> TriWizard Quest<i/></b><b>. Let your adventure begin! </b>

```python
# Start of game
def stage_1():
    print("""
Welcome to TriWizard Quest! Please enter your name to begin:
""")
    
    name = input() # Adding name
    if name == '':
        print('No worries, come back next time when you are ready!')
        return stage_1()  # Restarting the stage nothing was written
    else:    
        print(f'Hi, {name.capitalize()}! Let us start your adventure!')

    # Sorting Hat Section
    print("""
Before the quest begins, you must join a Hogwarts house. Each house has its unique traits. 
Are you ready to discover where you belong? Let's start the sorting!
""")

    # House values to put into for loop
    value = ['Courage', 'Ambition', 'Intelligence', 'Loyalty']
    # Putting a count for the loop
    count = 1
    
    # First question for the user
    print('What do you value most?')
    
    # FOR LOOP to iterate choices
    for choice in value:
             print(f'    {count} = {choice}') # Adding indentation for choices
             count += 1

    # WHILE LOOP: to assign house
    while True:
        chosen_value = input().lower() 
        if chosen_value in ['1', 'one', 'courage']: # Assigning to Gryffindor
            print('\nA brave and daring person like you deserves to be in... GRYFFINDOR!!')
            break
        elif chosen_value in ['2', 'two', 'ambition']: # Assigning to Slytherin
            print('\nAn ambitious and resourceful person like you deserves to be in... SLYTHERIN!!')
            break
        elif chosen_value in ['3', 'three', 'intelligence']:  # Assigning to Ravenclaw
            print('\nA wise and creative person like you deserves to be in... RAVENCLAW!!')
            break
        elif chosen_value in ['4', 'four', 'loyalty']: # Assigning to Hufflepuff
            print('\nA loyal and hardworking person like you deserves to be in... HUFFLEPUFF!!')
            break
        else:
            print('<Invalid choice. Please only choose between the given options>')
    
    # Starting TriWizard Tournament
    start = input(prompt = """
Now that you have your house sorted, let's start the TriWizard Tournament!
<Press enter to continue>
""")
    stage_2()
                
# Start of quest 1
def stage_2():
    print("""===============================================================================================
Welcome to Quest 1! The Dragon Challenge
Your task is to retrieve a golden egg guarded by a dragon, which will be randomly assigned to you.""")
    
    # Dragon list
    dragons = ['Hungarian Horntail', 'Swedish Short-Snout', 'Chinese Fireball', 'Common Welsh Green']
    
    # Adding dynamics to the game through randomizing the dragons to make it look like strategy varies per dragon
    import random
    randnum = random.randint(a = 0, 
                             b = 3)
    
    chosen_dragon = dragons[randnum]
    
    # Choosing strategy
    strategy1 = ['magic', 'agility', 'distraction'] # actual strategy
    
    # Adding a list of possible user input for control when user enters the wrong key
    control_input = ['magic', 'agility', 'distraction', '1', '2', '3', 'one', 'two', 'three'] 
    
    # Validating strategy and dragon
    while True: # Using while to make sure user enters the correct strategy
        chosen_strategy1 = input(prompt = """In order to win, choose between these strategies wisely:
    1 = Magic
    2 = Agility
    3 = Distraction
""").lower()  
            
        # Condition to make sure user enters the right keywords
        if chosen_strategy1 not in control_input:
            print('Please only choose between the three: 1 = Magic / 2 = Agility / 3 = Distraction\n')
            
        else:
           # Nested Conditional results for quest 1
            if chosen_strategy1 in ['magic', '1', 'one']:
                # Printing user choice
                print(f'\nYour chosen dragon is {chosen_dragon}. Your strategy is Magic. Ready to see the results?')
                quest1_result = input(prompt = '<Press enter to continue>\n')   
                
                # Result of choosing Magic which is the only right strategy
                print('Congratulations your strategy worked! You have retrieved the golden egg!')
                begin_quest2 = input(prompt = '<Press enter to continue to next quest>\n')
                stage_3() # Proceed to next stage
                break

            elif chosen_strategy1 in ['agility', '2', 'two'] :  # Providing a choice for agility

                 # Printing user choice to add different user experience
                print(f'\nYour chosen dragon is {chosen_dragon}. Your strategy is Agility. Ready to see the results?')
                quest1_result = input(prompt = '<Press enter to continue>\n')   
                
                # Result of choosing Agility, adding this section to provide user a chance
                print("""Unfortunately, agility is not working. Do you wish to change strategy?
<Type 'yes' or 'y' to change strategy, type anything else if you don't want to>""")
                
                while True: # Creating another loop in case they press enter
                    answer1 = input() # Giving chance to user to change strategy
                
                    if answer1.lower() == "": # Control if user suddenly input enter
                        print("""<You did not type anything, press enter or 'yes' or 'y' only if you want to change strategy.>""")
                        
                # Intentional loop for agility since user said they wanted to change strategy
                    elif answer1.lower() in ['yes' , 'y']: # Result if they want to change strategy
                        print('\nYou have chosen to change your strategy.')
                        break # Breaking second while loop
                    
                    else: # Stopping the loop if they had not chosen 'yes' or 'y'
                        print('\nYou have chosen not to change your strategy.')
                        break
                        
                if answer1.lower() not in ['yes' , 'y']:# Result if they do not want to change strategy
                        print('\nSorry, you failed the first quest and will not be able to proceed to the next.')
                        fail() # Stop game if they chose not to change strategy
                        break # Breaking second while loop
            
            else: 
                # Result for choosing Distraction which is the wrong strategy
                print(f'\nYour chosen dragon is {chosen_dragon}. Your strategy is Distraction. Ready to see the results?')
                quest1_result = input(prompt = '<Press enter to continue>\n')   
                
                # Wrong strategy
                print('Your strategy did not work.')
                fail()
                break

def stage_3(): #Start of quest 2
    
    print(f"""===============================================================================================
Now that you have the golden egg, you can prepare for the next quest!
What will be your next step?""")
    
    # Choosing strategy for the second quest
    strategy2 = ['Ask help from your fellow housemates', 'Study more magic', 'Believe in yourself']
    count = 1
    
    # Adding a list of possible user input for control when user enters the wrong key
    control_input2 = ['Ask help from your fellow housemates', 'Study more magic', 'Believe in yourself', 
                      '1', '2', '3', # For when user inputs a number
                      'one', 'two', 'three', # For when user inputs a word
                      'ask', 'help', 'from', 'your', 'fellow', 'housemates','ask help', # For keywords in option 1
                      'study', 'more', 'magic', 'study magic', # For keywords in option 2
                     'believe', 'in', 'yourself', 'believe yourself'] # For keywords in option 2
    
    # Using FOR loop to enumerate choices
    for choices2 in strategy2:
        print(f'    {count} = {choices2}') # Adding indentation
        count += 1
    
    while True:
        chosen_strategy2 = input().lower()
        
         # Control point for user input
        if chosen_strategy2 not in control_input2:
            print("""
<Please choose only between the three strategies (1-3)>
""")
        
        else:
            break # Breaking loop in user input control
        
    # Scenarios per strategy chosen for quest 2
    if chosen_strategy2 in ['1', 'one','ask', 'help', 'from', 'your', 'fellow', 'housemates','ask help', 'ask help from your fellow housemates']:
        chosenstrategy2 = 'Ask help from your fellow housemates'
        print(f"""
You have chosen to ask help from your fellow housemates!
You and your housemates strategized to tackle any water-related challenges.""")

        # Result of quest 2 for this strategy
        quest2_result = input(prompt = """
The next day, the quest was revealed to be to save your friends underwater!
<Press enter to continue>
""")

        print('Congratulations! With your strategy, you were able to breathe underwater and save your friends!')
        # end of quest 2
        begin_quest3 = input(prompt = '<Press enter to continue to next quest>\n')
        stage_4()

        # Result of quest 2 for this strategy
    elif chosen_strategy2 in ['2', 'two', 'study', 'more', 'magic', 'study magic','study more magic']:
        chosenstrategy2 = 'Study magic'
        print(f"""
You have chosen to study magic!
You were studying the whole night for every magic related to water.

The next day, the quest was revealed to be to save your friends underwater!
Before the quest, you're offered Gillyweed by a housemate.""")
        
        # Giving user another chance to win
        while True:
            answer3 = input(prompt = """
Will you trust your fellow housemate?
<Press 'yes' or 'y' to eat the Gillyweed, type anything else if you don't trust your housemate>
    """)
            # Nested Conditional to help user win
            if answer3.lower() == "": # Control if user suddenly input enter
                print("""<You did not type anything, please input 'yes' or 'y' if you want to eat the Gillyweed.
Type anything else if you do not want to.>""")

            # Result if they chose to trust the housemate
            elif answer3.lower() in ["yes", 'y']:
                print('Congratulations! The Gillyweed enabled you to breathe underwater and save your friends!')
                # end to quest 2
                begin_quest3 = input(prompt = '<Press enter to continue to next quest>\n')
                stage_4()
                break

            else:
                print('\nYou were not able to come up with a magic to save your friends. You should have trusted your fellow housemates.')
                fail() # end of game
                break # stop loop

    else:
        chosenstrategy2 = 'Believe in yourself'
        print(f"""
You have chosen to believe in yourself!
Confidence and trusting yourself are indeed important traits!""")

        # Result of quest 2 for this strategy
        quest2_result = input(prompt = """
The next day, the quest was revealed to be to save your friends underwater!
<Press enter to continue>
""")

        print('Unfortunately, confidence and trust in yourself would not let you breathe underwater and complete the quest.')
        fail() # End of game


# Start of last quest        
def stage_4():
    start_quest3 = input(prompt = """===============================================================================================
Welcome to The Maze Challenge! Your final quest!
Your task is to navigate the maze and find the Triwizard Cup. Be cautious of the obstacles!
<Press enter to continue>
""")

    print("""You enter the maze. At every turn, you face a decision:
    1 = Left
    2 = Right
    3 = Straight""")
    
    # Adding list depending on user input
    paths = ['left', 'right', 'straight']
    no_choice1 = ['1', '2', '3']
    word_choice2 = ['one', 'two', 'three']
    
    # Adding a list of possible user input for control when user enters the wrong key
    control_input3 = ['left', 'right', 'straight', '1', '2', '3', 'one', 'two', 'three']
    
    import random # Putting random for dynamic results on the correct path
    randnum2 = random.randint(a = 0, 
                             b = 1) # Right paths only for left and right to still control the randomness
    
    # Adding condition across three list to control what user will input
    correct_path = paths[randnum2]
    correct_no = no_choice1[randnum2]
    correct_word = word_choice2[randnum2]
    
    # Using while loop for maze navigation
    while True:
        choice = input(prompt = '\nWhich way do you go?\n').lower()
        
        # Control to make sure user enters the right words
        if choice not in control_input3:
            print("""
Not to worry, I know the maze is scary to navigate!
In order to move forward, please choose only between: 1 = Left / 2 = Right / 3 = Straight""")
            
        else: 
            break # Stops loop and proceeds to next condition
    
    # Condition of user chose the right pathj
    if choice == correct_path or choice == correct_no or choice == correct_word:
        print("""
Great choice! You were able to navigate through the maze successfully. Ahead of you lies the Triwizard Cup. 
You and Harry race towards it, but suddenly, Harry is yanked into the bushes, calling out for help!

Do you want to help him?""")
                
        while True: # Control for user input
            maze_challenge = input(prompt = """<Type 'yes' or 'y' to help him. Press any key to leave him be and run towards the cup!>
""")
                    
        # Control if user input blank
            if maze_challenge == "":
                print("""
<You did not type anything, please input 'yes' or 'y' if you want to help.
Type anything else if you do not want to.>""")

            # Challenge one for the maze before winning the game
            elif maze_challenge in ['yes', 'y']:
                print("""
By turning back to rescue Harry Potter, you averted an attack. Grateful, Harry lets you claim the cup!""")
                close_quest3 = input(prompt = '<Press enter to continue>\n')
                win()  # End of game succesfully
                break
                    
            else: # Condition for not choosing to save Harry
                print("""
You left Harry Potter and ran towards the cup, unfortunately, there was another attack that you were not 
able to anticipate. You should have been a better person and saved your fellow wizard.""")
                fail()
                break

    # Another chance for user to choose the right path
    else:
        print('It seems you are getting lost, do you want to retrace your steps and change paths?')
            
        while True:  # Adding loop for control point
            choice3 = input(prompt = """
<Type the path you want to change into: 1 = Left / 2 = Right / 3 = Straight
If you don't want to change paths choose your original path>
""").lower()
            
            # nested condition if the path chosen is correct
            if choice3 == correct_path or choice3 == correct_no or choice3 == correct_word:
                # Next scene if correct path is chosen
                print("""
Great choice! You were able to navigate through the maze successfully. Ahead of you lies the Triwizard Cup. 
You and Harry race towards it, but suddenly, Harry is yanked into the bushes, calling out for help!

Do you want to help him?""")

                while True: # Control for user input for next scene
                    maze_challenge = input(prompt = """<Type 'yes' or 'y' to help him. Press any key to leave him be and run towards the cup!>
""")

                    # Control if user input blank
                    if maze_challenge == "":
                        print("""
<You did not type anything, please input 'yes' or 'y' if you want to help.
Type anything else if you do not want to.>""")
                    
                    else: # Ending while loop since user put the correct word
                        break

                # Challenge one for the maze before winning the game
                if maze_challenge in ['yes', 'y']:
                    print("""
By turning back to rescue Harry Potter, you averted an attack. Grateful, Harry lets you claim the cup!""")
                    close_quest3 = input(prompt = '<Press enter to continue>\n')
                    win()  # End of game succesfully
                    break

                else: # Condition for not choosing to save Harry
                    print("""
You left Harry Potter and ran towards the cup, unfortunately, there was another attack that you were not 
able to anticipate. You should have been a better person and saved your fellow wizard.""")
                    fail()
                    break
                    
                    # condition if the user typed the wrong key in choosing paths
            elif choice3 not in control_input3: # control point for user input
                print("""Not to worry, I know the maze is scary to navigate!
In order to move forward, please choose only between: 1 = Left / 2 = Right / 3 = Straight.""")

            else: # Condition if they chose the wrong path
                print('\nYou are lost in the maze. Unfortunately, the challenge remains incomplete.')
                fail() # End of game
                break

# Establishing one win function            
def win():
    print("""================================================================================================
Congratulations! You have successfully completed each of the tasks and have proven yourself to be 
the Triwizard Champion!

Thank you for participating in Triwizard Quest!

Remember: We solemnly swear that we are up to no good ;)""")

# Establishing one fail function        
def fail():
    print('Game Over :(')
        
        
stage_1() 
```
